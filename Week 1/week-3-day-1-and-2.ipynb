```python
destination = input("Enter destination: ")
people = int(input("Enter number of people: "))
nights = int(input("Enter number of nights: "))
transport = float(input("Enter transport cost: "))
accommodation = float(input("Enter accommodation cost per night: "))

if destination == "Kathmandu" or destination == "Pokhara":
    total_transport = transport * people
    total_accommodation = accommodation * nights
    total_cost = total_transport + total_accommodation

    discount = 0

    if people >= 4:
        discount = discount + 10

    if nights >= 5:
        discount = discount + 5

    if discount > 15:
        discount = 15

    discount_amount = total_cost * discount / 100
    final_cost = total_cost - discount_amount
    cost_per_person = final_cost / people

    print("Destination:", destination)
    print("Number of people:", people)
    print("Number of nights:", nights)
    print("Transport cost:", total_transport)
    print("Accommodation cost:", total_accommodation)
    print("Total cost before discount:", total_cost)
    print("Discount amount:", discount_amount)
    print("Final trip cost:", final_cost)
    print("Cost per person:", cost_per_person)

else:
    print("Invalid destination")
```

```python
print("=" * 50)
print("       SALES COMMISSION CALCULATOR")
print("=" * 50)

number_of_employees = int(input("Enter number of employees: "))

total_commission = 0

for j in range(number_of_employees):
    print("\nEmployee", j + 1)

    name = input("Enter employee name: ")
    sales = float(input("Enter monthly sales (Rs.): "))

    if sales <= 10000:
        rate = 0.02
    elif sales <= 20000:
        rate = 0.05
    elif sales <= 50000:
        rate = 0.08
    else:
        rate = 0.12

    commission = sales * rate

    total_commission += commission

    print("Name:", name)
    print("Sales: Rs.", sales)
    print("Commission Rate:", rate * 100, "%")
    print("Commission: Rs.", commission)

print("\n" + "=" * 50)
print("Total Commission Paid: Rs.", total_commission)
print("=" * 50)
```

```python
strong_passwords = 0

for j in range(1, 11):
    password = input(f"Enter password {j}: ")

    criteria = 0
    missing = []

    if len(password) >= 8:
        criteria += 1
    else:
        missing.append("at least 8 characters")

    if any(char.isupper() for char in password):
        criteria += 1
    else:
        missing.append("an uppercase letter")

    if any(char.islower() for char in password):
        criteria += 1
    else:
        missing.append("a lowercase letter")

    if any(char.isdigit() for char in password):
        criteria += 1
    else:
        missing.append("a digit")

    if any(char in "!@#$%^&*" for char in password):
        criteria += 1
    else:
        missing.append("a special character")

    if criteria <= 2:
        strength = "Weak"
    elif criteria <= 4:
        strength = "Medium"
    else:
        strength = "Strong"
        strong_passwords += 1

    print(f"\nPassword {j}")
    print(f"Strength: {strength}")
    print(f"Criteria met: {criteria}/5")

    if missing:
        print("Missing:", ", ".join(missing))
    else:
        print("All criteria met!")

    print("-" * 40)

print(f"\nNumber of strong passwords: {strong_passwords}")
```

```python
def add(a, b):
    return a + b


def sub(a, b):
    return a - b


def mul(a, b):
    return a * b


def div(a, b):
    return a / b


num1 = float(input("Enter first number: "))
operator = input("Enter operation (+, -, *, /): ")
num2 = float(input("Enter second number: "))


if operator == "+":
    result = add(num1, num2)
elif operator == "-":
    result = sub(num1, num2)
elif operator == "*":
    result = mul(num1, num2)
elif operator == "/":
    if num2 != 0:
        result = div(num1, num2)
    else:
        result = "Cannot divide by zero"
else:
    result = "Invalid operation"


print("Result:", result)
```
