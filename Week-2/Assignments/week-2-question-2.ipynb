```python
# # Question 2 --- Tuition and scholarship calculator program

tuition = float(input("Enter tuition amount: "))
gpa = float(input("Enter GPA: "))
student_type = input("Enter student type (New or Continuing): ")
payment_method = input("Enter payment method (Full or Installment): ")
early_enrollment = input("Early enrollment (Yes or No): ")

if gpa >= 3.75:
    scholarship = 20
elif gpa >= 3.50:
    scholarship = 15
elif gpa >= 3.25:
    scholarship = 10
elif gpa >= 3.00:
    scholarship = 5
else:
    scholarship = 0

if payment_method == "Full":
    payment_discount = 5
else:
    payment_discount = 0

if early_enrollment == "Yes":
    early_discount = 3
else:
    early_discount = 0

total_discount = scholarship + payment_discount + early_discount

if total_discount > 30:
    total_discount = 30

discount_amount = tuition * total_discount / 100
final_tuition = tuition - discount_amount

if student_type == "New" and gpa >= 3.50:
    final_tuition = final_tuition - 300

print("Academic scholarship:", scholarship, "%")
print("Payment discount:", payment_discount, "%")
print("Early enrollment discount:", early_discount, "%")
print("Total discount:", total_discount, "%")
print("Final tuition:", final_tuition)
```
