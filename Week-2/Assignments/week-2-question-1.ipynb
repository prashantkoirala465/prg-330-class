```python
# Question 1 --- Attendance program 

total_classes = 16

attended = int(input("How many classes did you attend out of 16? "))

attendance = (attended / total_classes) * 100
late_classes = total_classes - attended

print("Attendance percentage:", attendance, "%")
print("Number of late classes:", late_classes)

if attendance >= 80 and late_classes <= 2:
    print("You can move to the next session.")
else:
    print("You cannot move to the next session.")
```
