In this notebook, I used basic NumPy features with the student data provided for the assignment. I first created a structured array with five students, then accessed and updated fields, added a new student, and finally used NumPy to do a few simple calculations with the full CSV file.

## 1. Import NumPy

I started by importing NumPy. I also printed the version to confirm that NumPy was working in the notebook.

```python
# Import NumPy
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Create a structured array

The assignment asks for fields for name, age, and height. I used the first five students from the provided `students.csv` file as the sample records.

```python
# Define the fields for the structured array
student_type = [
    ("name", "U30"),
    ("age", "i4"),
    ("height", "f4")
]

# Create the structured array with five students
students = np.array([
    ("Anisha Regmi", 21, 6.3),
    ("Amit Ghimire", 18, 6.1),
    ("Sujal Tamang", 19, 5.4),
    ("Amit Rai", 18, 5.5),
    ("Sabina Acharya", 25, 6.1)
], dtype=student_type)

print(students)
```

## 3. Access the Name field

A field in a structured array can be accessed by using the field name inside brackets.

```python
# Print all names
print(students["name"])
```

## 4. Update the age of the first student

The assignment asks to change the age of the first person to 30.

```python
# Update the first student's age
students["age"][0] = 30

print(students)
```

## 5. Add Nadine Smith

I created one new record using the same data type and then added it to the existing structured array.

```python
# Create the new record
new_student = np.array([
    ("Nadine Smith", 25, 5.9)
], dtype=student_type)

# Add the new student to the array
students = np.append(students, new_student)

print(students)
```

## 6. Load the full student CSV file with NumPy

For the additional part of the assignment, I used the full `students.csv` file provided by the TA. I used `np.genfromtxt()` because the file contains a header row and different data types.

```python
# Read the full CSV file
all_students = np.genfromtxt(
    "students.csv",
    delimiter=",",
    names=True,
    dtype=None,
    encoding="utf-8"
)

print("First five records:")
print(all_students[:5])

print("\nNumber of students:", len(all_students))
print("Fields:", all_students.dtype.names)
```

## 7. Basic NumPy analysis

I used simple NumPy functions to find the average, minimum, and maximum values for age and height.

```python
# Age calculations
average_age = np.mean(all_students["age"])
youngest_age = np.min(all_students["age"])
oldest_age = np.max(all_students["age"])

# Height calculations
average_height = np.mean(all_students["height"])
shortest_height = np.min(all_students["height"])
tallest_height = np.max(all_students["height"])

print("Average age:", round(average_age, 2))
print("Youngest age:", youngest_age)
print("Oldest age:", oldest_age)

print("\nAverage height:", round(average_height, 2))
print("Shortest height:", shortest_height)
print("Tallest height:", tallest_height)
```

## 8. Simple filtering

NumPy can also filter records using a condition. I checked how many students are age 25 and how many have a height of 6.0 or above.

```python
# Students who are 25 years old
age_25 = all_students[all_students["age"] == 25]

# Students with height 6.0 or above
height_6_or_more = all_students[all_students["height"] >= 6.0]

print("Number of students aged 25:", len(age_25))
print("Students with height 6.0 or above:", len(height_6_or_more))
```

## 9. Find the tallest students

The maximum height in the dataset is 6.3. I used that value as a condition to display all students with the tallest recorded height.

```python
# Find the tallest height
max_height = np.max(all_students["height"])

# Filter students who have the maximum height
tallest_students = all_students[all_students["height"] == max_height]

print("Tallest height:", max_height)
print("\nTallest students:")
print(tallest_students)
```

There are **24 students who are age 25** and **29 students with a height of 6.0 or above**. The tallest recorded height is **6.3**.
