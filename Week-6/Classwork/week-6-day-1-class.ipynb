## Check the time of numpy and python inbuilt list

```python
import numpy as np
import timeit
n = 1_000_000
my_list = list(range(n))
my_array=np.arange(n)

# List
list_time=timeit.timeit("[x*2 for x in my_list]", globals=globals(), number=50)

# NumPy
numpy_time=timeit.timeit("my_array*2", globals=globals(), number=50)

print(f"List average time: {list_time/10:.6f} seconds")
print(f"Numpy average time: {numpy_time/10:.6f} seconds")
print(f"\n Numpy is approximately {list_time/numpy_time:.2f}x faster")
```
